```python
DATA_DIR = "../data/ECG/raw"

files = {
    "MITBIH_TRAIN": "mitbih_train.csv",
    "MITBIH_TEST": "mitbih_test.csv",
    "PTBDB_NORMAL": "ptbdb_normal.csv",
    "PTBDB_ABNORMAL": "ptbdb_abnormal.csv"
}

data = {}

for name, filename in files.items():
    path = os.path.join(DATA_DIR, filename)

    df = pd.read_csv(path, header=None)
    data[name] = df

    print(f"\n===== {name} =====")
    print("Shape:", df.shape)
    print("Missing:", df.isnull().sum().sum())
    print("Label distribution:")
    print(df.iloc[:, -1].value_counts().sort_index())
```

```python
df = data["MITBIH_TRAIN"]

plt.figure(figsize=(14, 4))

for label in sorted(df.iloc[:, -1].unique()):
    sample = df[df.iloc[:, -1] == label].iloc[0, :-1]

    plt.plot(sample.values, label=f"Class {int(label)}")

plt.title("ECG Waveform - Different Classes")
plt.xlabel("Time Steps")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.show()
```
