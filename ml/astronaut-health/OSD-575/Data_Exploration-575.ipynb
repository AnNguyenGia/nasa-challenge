```python
import re
# Tách mã đối tượng và thời điểm lấy mẫu
def extract_sample_info(sample):
    match = re.match(
        r"(C\d+)_serum_([LR])([+-]\d+)",
        sample.strip()
    )

    if match:
        subject = match.group(1)
        day = int(match.group(3))

        return pd.Series([subject, day])

    return pd.Series([None, None])


temporal_df = cleaned_df.copy()

temporal_df[["Subject", "Mission_Day"]] = (
    temporal_df["Sample Name"].apply(extract_sample_info)
)

# Sắp xếp theo đối tượng và thời gian
temporal_df = temporal_df.sort_values(
    ["Subject", "Mission_Day"]
)

print(
    temporal_df[
        ["Sample Name", "Subject", "Mission_Day"]
    ].to_string(index=False)
)
 
```

```python
from scipy.stats import linregress

relation_results = []

for percent_col in immune_percent_cols:
    base_name = percent_col.replace("_percent", "")

    concentration_col = (
        base_name + "_concentration_picogram_per_milliliter"
    )

    if concentration_col in temporal_df.columns:
        x = temporal_df[concentration_col]
        y = temporal_df[percent_col]

        result = linregress(x, y)

        relation_results.append({
            "Feature": base_name,
            "Slope": result.slope,
            "Intercept": result.intercept,
            "R_squared": result.rvalue ** 2
        })

relation_df = pd.DataFrame(relation_results)

relation_df
```

R_squared = 1.0 ở cả 71 chỉ số: mô hình tuyến tính giải thích hoàn toàn sự biến thiên của Percent từ Concentration trong các mẫu hiện có.
Intercept đều rất gần 0 (khoảng \(10^{-8}\)): gần như không có hệ số chặn.
Slope khác nhau giữa các chỉ số: mỗi chỉ số có hệ số chuyển đổi riêng.

```python
from sklearn.metrics import mean_absolute_error

errors = []

for percent_col in immune_percent_cols:
    base_name = percent_col.replace("_percent", "")

    concentration_col = (
        base_name + "_concentration_picogram_per_milliliter"
    )

    if concentration_col in temporal_df.columns:
        x = temporal_df[concentration_col]
        y = temporal_df[percent_col]

        result = linregress(x, y)
        predicted = result.slope * x + result.intercept

        errors.append({
            "Feature": base_name,
            "MAE": mean_absolute_error(y, predicted)
        })

error_df = pd.DataFrame(errors)

print(error_df.describe())
```

Qua phân tích mối quan hệ giữa 71 thuộc tính Immune Concentration và Immune Percent, kết quả cho thấy cả 71 cặp đều có hệ số tương quan Pearson và Spearman bằng 1.0, hệ số xác định R² bằng 1.0. Đồng thời, MAE trung bình đạt khoảng 1.18 × 10⁻⁸, cho thấy sai số giữa giá trị thực tế và giá trị dự đoán theo quan hệ tuyến tính gần như bằng 0. Kết quả này cho thấy hai nhóm thuộc tính có quan hệ tuyến tính hoàn hảo trong bộ dữ liệu được khảo sát. Vì vậy, cần xem xét tránh sử dụng đồng thời cả hai nhóm làm đặc trưng đầu vào của mô hình Machine Learning nhằm hạn chế thông tin dư thừa.

```python

# Chọn các cột chỉ số đo thực tế
feature_cols = cmp_value_cols + immune_concentration_cols

# Tính số lượng outlier theo IQR
outlier_summary = []

for col in feature_cols:
    Q1 = clean_df[col].quantile(0.25)
    Q3 = clean_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((clean_df[col] < lower) |
             (clean_df[col] > upper)).sum()

    outlier_summary.append({
        "Feature": col,
        "Outlier_Count": count,
        "Outlier_Percentage": count / len(clean_df) * 100
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df = outlier_df.sort_values(
    by="Outlier_Count",
    ascending=False
)

outlier_df.head(20)
```

```python
outlier_details = []

for col in feature_cols:
    Q1 = clean_df[col].quantile(0.25)
    Q3 = clean_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    mask = (clean_df[col] < lower) | (clean_df[col] > upper)

    for idx in clean_df[mask].index:
        outlier_details.append({
            "Feature": col,
            "Sample": clean_df.loc[idx, "Sample Name"],
            "Subject": temporal_df.loc[idx, "Subject"],
            "Mission_Day": temporal_df.loc[idx, "Mission_Day"],
            "Value": clean_df.loc[idx, col]
        })

outlier_details_df = pd.DataFrame(outlier_details)

outlier_details_df.head(30)
```

```python
outlier_details = []

for col in feature_cols:
    Q1 = clean_df[col].quantile(0.25)
    Q3 = clean_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    mask = (clean_df[col] < lower) | (clean_df[col] > upper)

    for idx in clean_df[mask].index:
        value = clean_df.loc[idx, col]

        outlier_details.append({
            "Feature": col,
            "Sample": clean_df.loc[idx, "Sample Name"],
            "Subject": temporal_df.loc[idx, "Subject"],
            "Mission_Day": temporal_df.loc[idx, "Mission_Day"],
            "Value": value,
            "Outlier_Type": "Low" if value < lower else "High",
            "Lower_Bound": lower,
            "Upper_Bound": upper
        })

outlier_details_df = pd.DataFrame(outlier_details)

outlier_details_df.head(30)
```

Qua phương pháp IQR, dữ liệu xuất hiện các giá trị ngoại lệ ở cả nhóm chỉ số sinh hóa (CMP) và chỉ số miễn dịch (Immune). Các giá trị này phân bố ở nhiều đối tượng và thời điểm khác nhau. Do chưa có bằng chứng xác định đây là lỗi thu thập hoặc xét nghiệm, nghiên cứu giữ lại các giá trị ngoại lệ để tránh làm mất thông tin sinh học.

### Phân tích giá trị ngoại lệ (Outlier)
Nghiên cứu sử dụng phương pháp IQR (Interquartile Range) để phát hiện các giá trị ngoại lệ trong dữ liệu. Kết quả ghi nhận tổng cộng 105 lượt Outlier, bao gồm 96 giá trị High và 9 giá trị Low.
Các giá trị ngoại lệ xuất hiện ở cả nhóm chỉ số sinh hóa (CMP) và nhóm chỉ số miễn dịch (Immune). Trong đó, nhóm miễn dịch có số lượng ngoại lệ đáng chú ý, đặc biệt là TNF-α và IL-4 với 6 giá trị mỗi chỉ số.
